#TASK 4 — Data Visualization

##Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
```

##Loading Dataset

```python
df = pd.read_csv('/content/patient_feedback_cleaned.csv')
print(df.head())
```

## Visualizations

### 1. Sentiment Count

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Sentiment', data=df, palette='viridis')
plt.title('Sentiment Count')
plt.xlabel('Sentiment (0: Negative, 1: Positive)')
plt.ylabel('Count')
plt.show()
```

### 2. Satisfaction by Theme

```python
plt.figure(figsize=(10, 7))
sns.boxplot(x='Theme', y='Satisfaction', data=df, palette='pastel')
plt.title('Satisfaction Distribution by Theme')
plt.xlabel('Theme')
plt.ylabel('Satisfaction Score')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 3. Readmission Analysis

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Readmission', data=df, palette='magma')
plt.title('Readmission Count')
plt.xlabel('Readmission (0: No, 1: Yes)')
plt.ylabel('Count')
plt.show()

plt.figure(figsize=(10, 7))
sns.barplot(x='Theme', y='Readmission', data=df, palette='coolwarm')
plt.title('Readmission Rate by Theme')
plt.xlabel('Theme')
plt.ylabel('Average Readmission (0-1)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 4. Word Cloud

```python
all_feedback_text = ' '.join(df['Feedback'].dropna())

wordcloud = WordCloud(width=800, height=400, background_color='white').generate(all_feedback_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Word Cloud of Patient Feedback')
plt.show()
```
