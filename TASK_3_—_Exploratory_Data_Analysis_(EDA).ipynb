#TASK 3 — Exploratory Data Analysis (EDA)

##Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

##Loading Dataset

```python
df = pd.read_csv('/content/patient_feedback_cleaned.csv')

# Display the first few rows of the DataFrame
df.head()
```

### Dataset Information

```python
df.info()
```

### Descriptive Statistics

```python
df.describe(include='all')
```

### Sentiment Distribution

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Sentiment', data=df, palette='viridis')
plt.title('Distribution of Sentiment')
plt.xlabel('Sentiment (0: Negative, 1: Positive)')
plt.ylabel('Count')
plt.show()
```

### Satisfaction Distribution

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Satisfaction', data=df, palette='magma')
plt.title('Distribution of Satisfaction Scores')
plt.xlabel('Satisfaction Score')
plt.ylabel('Count')
plt.show()
```

### Theme Analysis

```python
plt.figure(figsize=(10, 7))
sns.countplot(y='Theme', data=df, order = df['Theme'].value_counts().index, palette='plasma')
plt.title('Distribution of Feedback Themes')
plt.xlabel('Count')
plt.ylabel('Theme')
plt.show()
```

### Sentiment by Theme

```python
plt.figure(figsize=(12, 8))
sns.countplot(y='Theme', hue='Sentiment', data=df, palette='viridis', order = df['Theme'].value_counts().index)
plt.title('Sentiment Distribution by Theme')
plt.xlabel('Count')
plt.ylabel('Theme')
plt.legend(title='Sentiment')
plt.show()
```

### Satisfaction by Theme

```python
plt.figure(figsize=(12, 8))
sns.boxplot(y='Theme', x='Satisfaction', data=df, palette='plasma', order = df['Theme'].value_counts().index)
plt.title('Satisfaction Scores by Theme')
plt.xlabel('Satisfaction Score')
plt.ylabel('Theme')
plt.show()
```
