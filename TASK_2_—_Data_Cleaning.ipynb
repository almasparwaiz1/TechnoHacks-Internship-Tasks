#TASK 2 — Data Cleaning

##Loading Dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/patient_feedback_raw_data.csv')

# Display the first few rows of the DataFrame
df.head()
```

### 1. Check Missing Values

```python
# Check for missing values
print("Missing values before cleaning:")
print(df.isnull().sum())
```

### 2. Remove Missing Rows (if any)

```python
# Remove rows with any missing values
df_cleaned = df.dropna()
print("\nMissing values after removing rows (should be 0):")
print(df_cleaned.isnull().sum())
print(f"Original rows: {len(df)}, Rows after dropping NA: {len(df_cleaned)}")
```

### 3. Remove Duplicate Rows

```python
# Remove duplicate rows
df_cleaned = df_cleaned.drop_duplicates()
print(f"\nRows after dropping duplicates: {len(df_cleaned)}")
```

### 4. Clean Text Column (Feedback)

```python
import re

# Function to clean text
def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r'[^a-z0-9\s]', '', text)  # Remove punctuation and special characters
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text

# Apply text cleaning to the 'Feedback' column
df_cleaned['Feedback'] = df_cleaned['Feedback'].apply(clean_text)
print("\nFirst 5 cleaned feedback entries:")
print(df_cleaned['Feedback'].head())
```

### 5. Saving Cleaned Dataset

```python
# Save the cleaned DataFrame to a new CSV file
df_cleaned.to_csv('patient_feedback_cleaned.csv', index=False)
print("\nCleaned dataset saved to 'patient_feedback_cleaned.csv'")

# Display the head of the cleaned dataframe
print("\nHead of the cleaned DataFrame:")
display(df_cleaned.head())
```
